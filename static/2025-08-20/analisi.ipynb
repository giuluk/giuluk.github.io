## Analisi del tiro da 3 punti nella NBA
Materiale di supporto per l'articolo: *[La rivoluzione del tiro da tre](https://giuluk.github.io/posts/)*

### Disclaimer
Il codice in questo notebook privilegia la **chiarezza** e la **semplicità** rispetto all’efficienza o all’eleganza, per renderlo accessibile anche a chi ha poca familiarità con Python.

```python
import pandas as pd

# Carica il file CSV
df = pd.read_csv("nba.csv")

# Ordina le stagioni dalla più vecchia alla più recente
df = df.sort_values(by="Season", ascending=True)
```

```python
# Grafico del numero medio di tiri da 3 punti tentati per squadra a partita
ax1 = df.plot(
    x="Season",
    y="3PA",
    color='darkorange',
    legend=False,
    grid=True,
    figsize=(10, 5.6),
    title="Numero medio di tiri da 3 punti a partita per squadra"
)

ax1.set_xlabel("Stagione NBA")
ax1.set_ylabel("Numero di tiri a partita");
```

```python
# Grafico del numero medio di tiri totali e da 3 punti tentati per squadra a partita
ax2 = df.plot(
    x="Season",
    y=["FGA", "3PA"],
    color=['blue', 'darkorange'],
    grid=True,
    figsize=(10, 5.6),
    title="Numero medio di tiri totali e da 3 punti a partita per squadra"
)

ax2.set_xlabel("Stagione NBA")
ax2.set_ylabel("Numero di tiri a partita");
```

```python
# Calcolo dell'incidenza dei tiri da 3 punti sul totale
df["Incidenza_3PA"] = (df["3PA"] / df["FGA"]) * 100

# Grafico dell'incidenza dei tiri da 3 punti sul totale
ax3 = df.plot(
    x="Season",
    y="Incidenza_3PA",
    color='green',
    legend=False,
    grid=True,
    figsize=(10, 5.6),
    title="Incidenza dei tiri da 3 punti sul totale"
)

ax3.set_xlabel("Stagione NBA")
ax3.set_ylabel("Percentuale")
ax3.axhline(y=50, color='red', linestyle='--');
```
